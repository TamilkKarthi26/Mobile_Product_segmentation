```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/mobile_reviews_cleaned.csv")

print(df.shape)
df.head()
```

```python
plt.figure(figsize=(10,5))

sns.countplot(
    data=df,
    x="brand",
    order=df["brand"].value_counts().index
)

plt.title("Brand Distribution")
plt.xticks(rotation=45)
plt.show()
```

The chart shows the number of reviews received by each mobile brand. Brands with higher review counts have greater customer engagement.

```python
plt.figure(figsize=(6,4))

sns.countplot(
    data=df,
    x="sentiment"
)

plt.title("Sentiment Distribution")
plt.show()
```

Most reviews are Positive, indicating overall customer satisfaction.

```python
plt.figure(figsize=(8,5))

sns.histplot(
    df["rating"],
    bins=5,
    kde=True
)

plt.title("Rating Distribution")
plt.show()
```

Ratings are concentrated around middle-to-high values, suggesting generally favorable product experiences.

```python
plt.figure(figsize=(10,5))

sns.countplot(
    data=df,
    x="country",
    order=df["country"].value_counts().index
)

plt.title("Country-wise Reviews")
plt.xticks(rotation=45)
plt.show()
```

Review activity varies across countries, showing differences in market participation.

```python
brand_rating = (
    df.groupby("brand")["rating"]
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(10,5))

brand_rating.plot(kind="bar")

plt.title("Average Rating by Brand")
plt.ylabel("Average Rating")
plt.show()
```

Some brands consistently receive higher ratings, indicating stronger customer satisfaction.
